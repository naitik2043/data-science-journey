# MNIST Neural Network

## Let's Start with the MNIST Dataset

Previously, we visualized the **MNIST dataset** and saw that each image is a **28 × 28 grayscale image** representing a handwritten digit from **0 to 9**.

Now, instead of using traditional Machine Learning algorithms, we'll build an **Artificial Neural Network (ANN)** that can learn complex patterns and classify handwritten digits with high accuracy.

```python
from tensorflow.keras.datasets import mnist

# Load the MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

print("Training Images:", x_train.shape)
print("Training Labels:", y_train.shape)
```

### Interpretation

- **60,000** images are used for training.
- **10,000** images are used for testing.
- Each image has a size of **28 × 28 pixels**.
- Labels range from **0 to 9**, representing the actual handwritten digit.

---

## Why Do We Need a Neural Network?

A simple algorithm like the **Perceptron** can only solve **linearly separable** problems.

Handwritten digits are much more complex because:

- Every person's handwriting is different.
- Digits may have different shapes, sizes, and thicknesses.
- Images contain **784 pixel values (28 × 28)** that need to be analyzed together.

A **Neural Network** learns these complex patterns automatically and makes accurate predictions.

---

## Building the Neural Network

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense

model = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(128, activation="relu"),
    Dense(10, activation="softmax")
])
```

### Architecture

```
28 × 28 Image
        │
        ▼
    Flatten Layer
(28×28 → 784 Features)
        │
        ▼
Dense Layer (128 Neurons)
      ReLU
        │
        ▼
Output Layer (10 Neurons)
    Softmax
```

---

## Compile and Train the Model

```python
model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=32
)
```

During training, the Neural Network:

- Makes predictions.
- Calculates the error (loss).
- Updates weights using backpropagation.
- Repeats the process for multiple epochs.

---

## Evaluate the Model

```python
test_loss, test_acc = model.evaluate(x_test, y_test)

print("Test Accuracy:", test_acc)
```

The model is evaluated on unseen test images to measure how well it has learned.

---

## Key Components

| Component | Purpose |
|-----------|---------|
| `Sequential()` | Creates a sequential neural network |
| `Input()` | Defines the input image shape |
| `Flatten()` | Converts a 28×28 image into a 784-element vector |
| `Dense()` | Fully connected neural network layer |
| `ReLU` | Hidden layer activation function |
| `Softmax` | Converts outputs into class probabilities |
| `Adam` | Optimizer that updates weights efficiently |
| `Categorical Crossentropy` | Loss function for multi-class classification |

---

## Summary

- MNIST is a **multi-class classification** problem with **10 digit classes (0–9)**.
- Images are first **normalized** and labels are **one-hot encoded** before training.
- The **Flatten layer** converts image data into a 1D vector.
- Hidden layers learn important features using the **ReLU activation function**.
- The **Softmax layer** predicts the probability for each digit class.
- The model is trained using **Adam Optimizer** and **Backpropagation** to improve accuracy over multiple epochs.

This is one of the simplest yet most powerful examples of an **Artificial Neural Network (ANN)** for image classification.

---

````python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Input
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical # This is used for one hot encoding
````

````python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
````

````python
x_train = x_train/255.0
x_test = x_test/255.0
````

````python
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
````

````python
model = Sequential([
    # Flatten(input_shape=(28, 28)),
    Input(shape=(28, 28)),          
    Flatten(),
    Dense(128, activation="relu"),
    Dense(10, activation="softmax")
])
````

````python
model.compile(
    optimizer = "adam",
    loss="categorical_crossentropy",
    metrics=['accuracy'])
````

````python
model.fit(x_train, y_train, epochs=5, batch_size=32)
````

````python
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"Test accuracy is {test_acc}")
````
